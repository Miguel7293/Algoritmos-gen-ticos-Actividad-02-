## 1. Datos y particiones independientes
300 puntos uniformes en [-1, 1]². La clase es 1 si ambas coordenadas tienen el mismo signo.
Dividimos estratificadamente en 60 % entrenamiento, 20 % validación y 20 % prueba.
La validación guía al AG; el test se consulta solamente al cerrar la búsqueda.


```python
import random
import warnings
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPClassifier
from sklearn.exceptions import ConvergenceWarning

SEED = 10
datos_rng = np.random.RandomState(SEED)
X = datos_rng.uniform(-1, 1, (300, 2))
y = (X[:, 0] * X[:, 1] > 0).astype(int)
idx_dev, idx_test = train_test_split(np.arange(len(y)), test_size=0.20, stratify=y, random_state=SEED)
idx_train, idx_val = train_test_split(idx_dev, test_size=0.25, stratify=y[idx_dev], random_state=SEED)
X_train, y_train = X[idx_train], y[idx_train]
X_val, y_val = X[idx_val], y[idx_val]
X_test, y_test = X[idx_test], y[idx_test]
assert set(idx_train).isdisjoint(idx_val) and set(idx_dev).isdisjoint(idx_test)
print(f'Entrenamiento: {len(y_train)} | Validación: {len(y_val)} | Prueba: {len(y_test)}')
```

## 2. Cromosoma e inicialización
`capa1 ∈ {2,4,8}`, `capa2 ∈ {0,2,4}`, `activacion ∈ {relu,tanh}`.
El valor cero elimina la segunda capa: se exploran 18 arquitecturas posibles.
Inicializamos 8 individuos al azar y permitimos repeticiones; la caché evita reentrenarlos.


```python
ESPACIO = {'capa1': [2, 4, 8], 'capa2': [0, 2, 4], 'activacion': ['relu', 'tanh']}
rng = random.Random(SEED)
cache = {}
def crear_individuo():
    return {g: rng.choice(valores) for g, valores in ESPACIO.items()}

def capas(ind):
    return (ind['capa1'],) if ind['capa2'] == 0 else (ind['capa1'], ind['capa2'])

def crear_modelo(ind):
    return MLPClassifier(hidden_layer_sizes=capas(ind), activation=ind['activacion'],
                         solver='lbfgs', max_iter=2000, max_fun=30000, random_state=SEED)

def entrenar(modelo, datos, etiquetas):
    # Se registran y reportan las advertencias; no se ocultan fallos de convergencia.
    with warnings.catch_warnings(record=True) as avisos:
        warnings.simplefilter('always', ConvergenceWarning)
        modelo.fit(datos, etiquetas)
    convergencia = [a for a in avisos if issubclass(a.category, ConvergenceWarning)]
    for aviso in avisos:
        if not issubclass(aviso.category, ConvergenceWarning):
            warnings.warn(str(aviso.message), aviso.category)
    return len(convergencia)
```

## 3. Aptitud
Fitness = exactitud en los 60 puntos de validación, tras entrenar con los 180 de entrenamiento.
Semilla y optimizador son iguales para todos. Esta búsqueda puede sobreajustarse a validación;
por eso informamos también la prueba independiente al final.


```python
def evaluar(individuo):
    key = tuple(individuo[g] for g in ESPACIO)
    if key not in cache:
        modelo = crear_modelo(individuo)
        avisos = entrenar(modelo, X_train, y_train)
        cache[key] = {'fitness': float(modelo.score(X_val, y_val)), 'avisos': avisos}
    return cache[key]['fitness']
```

## 4. Selección, cruce y mutación
Conservamos la selección del original: tomar dos padres diferentes entre los cinco mejores.
Cruce uniforme: cada gen del hijo procede de uno de los padres con igual probabilidad.
Mutación de 0.20 por gen, cambiándolo a un valor distinto y válido.
Elitismo: dos mejores pasan mediante copias, sin mutar los padres.


```python
def seleccionar(poblacion):
    ordenados = sorted(poblacion, key=evaluar, reverse=True)
    return rng.sample(ordenados[:5], 2)

def cruzar(padre1, padre2):
    return {g: rng.choice([padre1[g], padre2[g]]) for g in ESPACIO}

def mutar(individuo, tasa=0.20):
    hijo = individuo.copy()
    for g, valores in ESPACIO.items():
        if rng.random() < tasa:
            hijo[g] = rng.choice([v for v in valores if v != hijo[g]])
    return hijo
```

## 5. Evolución y terminación
Ejecutamos 10 generaciones con 8 individuos y 2 élites. El historial incluye la generación 0
y la generación final 10. El mejor y el promedio pueden coincidir si hay cromosomas repetidos.


```python
def ejecutar_ag(tamano=8, generaciones=10, elite=2):
    assert tamano >= 5 and 0 < elite < tamano and generaciones >= 0
    rng.seed(SEED)
    cache.clear()
    poblacion = [crear_individuo() for _ in range(tamano)]
    historial = []
    for generacion in range(generaciones + 1):
        poblacion.sort(key=evaluar, reverse=True)
        valores = [evaluar(ind) for ind in poblacion]
        historial.append({'generacion': generacion, 'mejor': max(valores),
                          'media': float(np.mean(valores)), 'peor': min(valores)})
        print(f'Generación {generacion:02d}: mejor={max(valores):.4f}, media={np.mean(valores):.4f}')
        if generacion == generaciones:
            break
        nueva = [ind.copy() for ind in poblacion[:elite]]
        while len(nueva) < tamano:
            padre1, padre2 = seleccionar(poblacion)
            nueva.append(mutar(cruzar(padre1, padre2)))
        poblacion = nueva
    return poblacion, historial

poblacion, historial = ejecutar_ag()
mejor = poblacion[0].copy()
assert all(historial[i+1]['mejor'] >= historial[i]['mejor'] - 1e-12 for i in range(len(historial)-1))
print('Mejor arquitectura observada:', mejor, '| Validación:', evaluar(mejor))
print('Arquitecturas evaluadas:', len(cache), 'de 18')
avisos_busqueda = sum(v['avisos'] for v in cache.values())
print('Advertencias de convergencia durante la búsqueda:', avisos_busqueda)
```

## 6. Visualización del ciclo y comparación de la población final


```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
for campo, estilo in [('mejor', 'o-'), ('media', 's--'), ('peor', 'x:')]:
    axes[0].plot([h['generacion'] for h in historial], [h[campo] for h in historial], estilo, label=campo)
axes[0].set(xlabel='Generación', ylabel='Exactitud de validación', title='Evolución de arquitecturas')
axes[0].legend(); axes[0].grid(alpha=0.3)
casos = [poblacion[0], poblacion[len(poblacion)//2], poblacion[-1]]
axes[1].bar(['Mejor', 'Intermedia', 'Peor'], [evaluar(c) for c in casos], color=['#086b62', '#7798ab', '#d1b17c'])
axes[1].set(ylim=(0, 1.05), ylabel='Exactitud de validación', title='Población final (puede tener duplicados)')
plt.tight_layout(); plt.show()
for nombre, ind in zip(['Mejor', 'Intermedia', 'Peor'], casos):
    print(nombre, ind, 'Validación:', evaluar(ind))
```

## 7. Referencia fija y prueba final
Referencia: una capa de 4 neuronas y activación tanh, fijada antes de consultar el test.
Para prueba final, reentrenamos referencia y ganador con los 240 puntos de desarrollo
(entrenamiento + validación). Evaluamos ambos en los 60 puntos de test; no volvemos a seleccionar.


```python
referencia_genes = {'capa1': 4, 'capa2': 0, 'activacion': 'tanh'}
evaluaciones_busqueda = len(cache)
referencia_val = evaluar(referencia_genes)
ganador_val = evaluar(mejor)
referencia = crear_modelo(referencia_genes)
modelo_final = crear_modelo(mejor)
avisos_base = entrenar(referencia, X[idx_dev], y[idx_dev])
avisos_final = entrenar(modelo_final, X[idx_dev], y[idx_dev])
base_test = float(referencia.score(X_test, y_test))
ag_test = float(modelo_final.score(X_test, y_test))
resultado = {'ejemplo': 'Neuroevolución', 'semilla': SEED, 'n_train': len(y_train),
             'n_validacion': len(y_val), 'n_test': len(y_test), 'referencia_validacion': referencia_val,
             'mejor_validacion': ganador_val, 'referencia_test': base_test, 'ag_test': ag_test,
             'ganador': mejor, 'capas_ocultas': list(capas(mejor)),
             'evaluaciones_unicas': evaluaciones_busqueda,
             'avisos_convergencia_busqueda': avisos_busqueda,
             'avisos_convergencia_referencia_final': avisos_base,
             'avisos_convergencia_ganador_final': avisos_final, 'historial': historial}
print(f'Referencia: validación={referencia_val:.2%}, test={base_test:.2%}')
print(f'AG: validación={ganador_val:.2%}, test={ag_test:.2%}')
print(f'Advertencias en reentrenamiento: referencia={avisos_base}, ganador={avisos_final}')
```

## Para explicar en la evaluación oral
1. La arquitectura evoluciona; los pesos se entrenan con L-BFGS en cada evaluación.
2. `capa2=0` elimina una capa: el AG sí cambia la arquitectura.
3. Identifica los siete pasos del ciclo en las secciones anteriores, incluido el criterio de parada.
4. Una exactitud alta en validación no garantiza el mismo resultado en prueba.
5. Se exploran solo 18 arquitecturas posibles con una semilla. No es una demostración de optimalidad
   ni una comparación exhaustiva con búsqueda aleatoria o búsqueda en rejilla.
